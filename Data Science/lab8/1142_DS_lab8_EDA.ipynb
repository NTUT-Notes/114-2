```python
# Joint Plot
# Seaborn聯合圖包括三個圖；在這三個圖中，第一個圖將顯示雙變數圖，顯示因變數將如何隨變數變化。
# 第二個聯合圖放置在雙變數圖的頂部，顯示該因變數的分佈。
# 第三個聯合圖放在雙變數圖的右邊，顯示該因變數的分佈。
# 將單變數和雙變數一起畫在一張圖上是非常有幫助的。單變數分析集中在一個變數上，它將描述總結和顯示資料模式。

sns.jointplot(x='Fare',y='Age', data=df)
```

```python
### Distribution Plot  ( KDE: kernel density estimate)
sns.distplot(df["Fare"],kde=True) # with the kde
```

```python
###Box Plot
sns.boxplot(x='Pclass', y='Age', data=df)
```

```python
### Swarm plot  分簇散點图
sns.swarmplot(x='Pclass', y='Age', data=df)
```

```python
# Stripplot 分布散点图
sns.stripplot(x='Pclass', y='Fare', data=df, jitter=True)
```

```python
### Count plot
sns.countplot(x='Sex', data=df)
```

```python
### Facet Grid  (一面多圖，以網格顯示圖面)
###
g = sns.FacetGrid(data=df, col='Sex')
g.map(sns.distplot, 'Age', kde=True)
```

```python
g = sns.FacetGrid(data=df, row='Survived', col='Pclass')
g.map(sns.distplot, 'Age', kde=True)
g.savefig("facet_plot1.png")
```

```python
g = sns.FacetGrid(data=df, row='Survived', col='Pclass')
g.map(sns.scatterplot, 'Age', 'Fare')
g.savefig("facet_plot2.png")
```

```python
# Heat maps
# for a heat map, both indexing and correaltions needs to be calculated first

titanic=pd.merge(df[['PassengerId','Survived']], df[['PassengerId','Pclass']], on='PassengerId', suffixes=('_Survive', '_Pclass'))

tc = titanic.corr()

sns.heatmap(tc,cmap='coolwarm')

plt.title('Titanic correlation coefficient')
```

```python
# Look at survival rate by sex (性別存活率)
df.groupby('Sex')[['Survived']].mean()
```

```python
# Look at survival rate by Pclass (艙別等級)
df.groupby('Pclass')[['Survived']].mean()
```

```python
# Look at survival rate by sex and class 觀察女性男性坐不同艙別的存活機會 
# Pandas套件的Pivot Table樞紐分析表，就能夠將欄位資料透過交叉比對的方式，進行群組、匯總及統計，幫助資料分析人員快速解讀資料。
df.pivot_table('Survived', index = 'Sex', columns = 'Pclass')
```

```python
# Look at survival rate by sex, age and class
age = pd.cut(df['Age'],[0, 18, 80])
df.pivot_table('Survived', index = ['Sex', age], columns ='Pclass', observed=False)
```

## 練習: 使用EDA 工具，分析Boston Housing Price 資料集 "Boston Housing")

```python
# https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_boston.html
# from sklearn.datasets import load_boston
# boston_dataset = load_boston()

data_BH = pd.read_csv('./BostonHousing.csv',index_col=0)
```

```python
data_BH.head()
```

## Attribute Information (in order):
- CRIM     per capita crime rate by town (犯罪率)
- ZN       proportion of residential land zoned for lots over 25,000 sq.ft. （居住使用土地比例）
- INDUS    proportion of non-retail business acres per town （非零售業土地使用比例）
- CHAS     Charles River dummy variable (= 1 if tract bounds river; 0 otherwise) (是否鄰近Charles River/河邊宅)
- NOX      nitric oxides concentration (parts per 10 million) (氮氧化合物濃度/空汙指標)
- RM       average number of rooms per dwelling （房子有幾間房間）
- AGE      proportion of owner-occupied units built prior to 1940 （屋齡）
- DIS      weighted distances to five Boston employment centres （到波士頓商業中心的距離）
- RAD      index of accessibility to radial highways (環狀高速公路的可觸指標/交通便利性)
- TAX      full-value property-tax rate per 10,000 (財產稅占比(每一萬美元)/財產稅率)
- PTRATIO  pupil-teacher ratio by town (生師比/教育)
- B        1000(Bk - 0.63)^2 where Bk is the proportion of blacks by town (黑人比例)
- LSTAT    lower status of the population (中低收入戶佔當地居住人口的比例/低社經人口比例)
- MEDV     Median value of owner-occupied homes in $1000's (以千美元計的房價中位數/房價)

```python
data_BH.info()
```

```python
profile_BH = ProfileReport(data_BH, title="EDA analysis for Boston Housing Price")
profile_BH.to_file("Boston_Housing.html")
```

```python
boston = pd.DataFrame(data_BH)
boston.head()
```

```python
sns.distplot(boston['medv'])
```

```python
correlation_matrix = boston.corr().round(2)
#  annot = True 讓我們可以把數字標進每個格子裡
# 觀察每個變數之間的關係，透過相關係數去觀察有哪些特徵變數和目標變數有較高的相關性
sns.heatmap(data=correlation_matrix, annot = True)
```

###  相關係數表格可以觀察--> 跟MEDV（房價）高度相關的是LSTAT（中低收入戶佔當地居住人口的比例）和RM（房子有幾間房間）這兩個變數

```python
# 設定整張圖的長寬
plt.figure(figsize=(20, 5))
features = ['lstat','rm']
target = boston['medv']
for i, col in enumerate(features):
 # 排版1 row, 2 columns, nth plot：在jupyter notebook上兩張並排 
 plt.subplot(1, len(features) , i+1)
 # add data column into plot
 x = boston[col]
 y = target
 plt.scatter(x, y, marker='o')
 plt.title(col)
 plt.xlabel(col)
 plt.ylabel('MEDV')
```
